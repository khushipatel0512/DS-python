```python
#1. Write a python program to sum of the first n positive integers.
n=int(input("enter the positive n interger"))
sum=0
for i in range(1,n+1):
    sum=sum+i
    print("sum of first",n,"positive interger is",sum)
```

```python
#13.Write a Python program to sort a dictionary (ascending /descending) by value.
color_dict = {'red': 3, 'green': 1, 'blue': 5, 'yellow': 2}
print("Original dictionary:", color_dict)

# Sort in ascending order by value
sorted_dict_asc = dict(sorted(color_dict.items(), key=lambda item: item[1]))
print("Ascending order:", sorted_dict_asc)

# Sort in descending order by value
sorted_dict_desc = dict(sorted(color_dict.items(), key=lambda item: item[1], reverse=True))
print("Descending order:", sorted_dict_desc)
```
